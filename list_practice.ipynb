TUPLES: if any item dont want to modify/edit at any time then use tuple

```python
days = ("sunday", "monday", "tuesday", "wednusday", "thursday", "friday", "saturday")
print(days)
print(type(days))
```

```python
S = ("sunday")               
E = (23,43)                         #tuple will create in () and items should be two with seperated commas (,)
print(type(E))
print(S)
print(type(S))
```

```python
empty = ()              #we have empty tuple
print(empty)
print(type(empty))
```

```python
numbers = (2,43,6,57,8,45,11)
print(numbers)                          #tuple will sort the numbers
items = tuple(sorted(numbers))
print(items)
```

UNION/INTERSECTION/DIFFERENTS: 

```python
batch_a = {"Rahul", "Priya", "Amit", "Sneha"}
batch_b = {"Amit", "Vikram", "Sneha", "Neha"}       
all_students = batch_a | batch_b                #intersection gives the all values of batch_a, batch_b
print(all_students)
print(len(all_students))
```

```python
batch_a = {"Rahul", "Priya", "Amit", "Sneha"}
batch_b = {"Amit", "Vikram", "Sneha", "Neha"}       
all_students = batch_a & batch_b                #union gives the same values of batch_a, batch_b
print(all_students)
print(len(all_students))
```

```python
batch_a = {"Rahul", "Priya", "Amit", "Sneha"}
batch_b = {"Amit", "Vikram", "Sneha", "Neha"}       
all_students = batch_a - batch_b                #subtraction gives uniques values of batch_a
print(all_students)
print(len(all_students))
```
